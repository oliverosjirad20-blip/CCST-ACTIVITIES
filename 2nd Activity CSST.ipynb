Problem: A student received the following grades:

```r
quiz <- 85
lab <- 90
project <- 88
exam <- 80

quiz_grade = 85 * .2
lab_grade = 90 * .25
project_grade = 88 * .25
exam_grade = 80 * .3

final_grade = quiz_grade + lab_grade + project_grade + exam_grade

cat("Final Grade:", final_grade)
```

7. Creating a Function
Instead of repeatedly writing the same code, create a function called

```r
classify_grad <- function(grade) {
  if (grade > 100 || grade < 0) {
    print("The grade is invalid")
  } else if (grade == 100) {
    print("Excellent")
  } else if (grade >= 89) {
    print("Very Good")
  } else if (grade >= 84) {
    print("Good")
  } else if (grade >= 75) {
    print("Passed")
  } else {
    print("Failed")
  }
}


classify_grad(90)
classify_grad(87)
classify_grad(75)
classify_grad(73)
classify_grad(84)
```

8. Creating a Data Frame

```r
students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95, 72, 88, 79, 68),
  Grade = c(90, 70, 85, 77, 65)
)
students
```

9. Adding a Classification Column


```r
students$Performance <- ifelse(
  students$Grade >= 80,
  "Good",
  ifelse (
    students$Grade >= 75,
    "Passed",
    "Failed"
  )
)
students
```

10. Creating a Simple Based Rule-Based System

```r
student_risk <- function(attendance,grade)
if (attendance < 75 && grade < 75) {
    return("High Risk");
} else if (attendance < 85 && grade < 80) {
    return("Moderate Risk");
} else {
    return("Low Risk");
}
```

11. Apply the rules to Multiple Students


```r
students$Risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
)
students
```

12. Count The Reuslts




```r
table(students$Risk)
```

13. Create Visualization

```r
risk_counts <- table (students$Risk)

barplot(
  risk_counts,
  main = "Student Risk Classificiation",
  xlab = "Risk Level",
  ylab = "Number of Students"
)
```

14. Add a Recommendation

```r
recommendation <- function(risk){
  if (risk == "High Risk") {
    return("Immediate academic invtervention")
  }else if (risk == "Moderate Risk"){
    return("Academic monitoring")
  }else{
    return("Continue regular monitoring")
  }
}
students$Recommnedation <- mapply(
  recommendation,
  students$Risk
)
students
```
